```python
!pip install requests pandas tqdm

import requests, pandas as pd, time
from tqdm import tqdm

# ---- adjustable parameters ----
query = "machine learning lithium ion battery"
max_results = 100          # try 50 first if connection is weak
sleep_time = 2             # wait between requests
# --------------------------------

papers = []
base_url = "https://api.semanticscholar.org/graph/v1/paper/search"

for offset in tqdm(range(0, max_results, 20)):
    params = {
        "query": query,
        "limit": 20,
        "offset": offset,
        "fields": "title,year,authors,url,venue,abstract,citationCount"
    }
    try:
        r = requests.get(base_url, params=params, timeout=20)
        if r.status_code == 200:
            data = r.json()
            papers.extend(data.get("data", []))
        elif r.status_code == 429:
            print("⚠️ Too many requests — waiting 30 s")
            time.sleep(30)
        else:
            print(f"Error {r.status_code}")
        time.sleep(sleep_time)
    except Exception as e:
        print("Request failed:", e)

# convert to DataFrame
df = pd.DataFrame(papers)
df.to_csv("battery_ml_papers.csv", index=False)
print(f"✅ Saved {len(df)} papers to battery_ml_papers.csv")
df.head()
```
